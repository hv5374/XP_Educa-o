```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

CRIAR MATRIZ USUARIO-ITEM

__________________________________________________________________________________________________________________

Representar na tabela, linhas(Usuários) e colunas(filmes) valores é a avaliação dos filmes

```python
user_movie_matrix = ratings.pivot(index='userId', columns='movieId', values='rating')
```

**SUBSTITUIR NaN**

__________________________________________________________________________________________________________________

Substituimos NaN por 0 para calcular a similaridade entre usuários, precisa que ***todos os valores da matriz sejam numéricos***.

Motivos principais

- Evitar erros nos calculos
- Representar ausencia de avaliação
- Padronização dos dados

```python
#Substituir valor NaN por 0

user_movie_matrix = user_movie_matrix.fillna(value=0)
```

```python
user_movie_matrix
```

SIMILARIDADE DO COSSENO
__________________________________________________________________________________________________________________

Mede quão parecidos são dois itens ou usuarios com base em seus padrões de comportamento

Imagine dois usuarios que avaliam filmes, comparamos o padrão de avaliação e vemos se tem gostos parecidos. Se deram notas altas no mesmo filme, tem provavelmente gostos parecidos.

A similaridade do cosseno compara os angulos entre os dois dados. Quanto meno o angulo entre os vetores, mais parecidos são os gostos.

```python
user_similarity = cosine_similarity(user_movie_matrix)
```

**CRIAR DATAFRAME COM AS SIMILARIDADES**

```python
user_sim_df = pd.DataFrame(data=user_similarity, 
             index=user_movie_matrix.index, 
             columns=user_movie_matrix.index)
```

**Analisando usuarios mais similares**

```python
similar_users = user_sim_df[5].sort_values(ascending=False).head(10)
```

**Visualizar notas dos usuários**

```python
user_movie_matrix.loc[similar_users.index]
```

**Desenvolvendo um sistema de recomendação de filmes**


__________________________________________________________________________________________________________________

```python
def recommend_movies(user_id, num_recommendations=5):

    # Filtro dos 10 primeiros usuários
    similar_users = user_sim_df[user_id].sort_values(ascending=False).head(10)

    #Criar matriz de avaliações
    similar_user_ratings = user_movie_matrix.loc[similar_users.index]

    # Multiplicar avaliações pela similaridade, para dar maior peso aos usuarios mais similares
    weighted_user = similar_user_ratings.T.dot(similar_users.index)

    # Remover filmes já avaliados pelo usuário
    unseen_movies = user_movie_matrix.loc[user_id] == 0
    recommended_movies = weighted_user[unseen_movies].sort_values(ascending=False).head(num_recommendations)

    #retornar nome dos filmes recomendados
    return movies[movies['movieId'].isin(recommended_movies.index)][['title', 'movieId']]
```

**Aplicação do Sistema de recomendação baseado em usuários similares**

```python
user_id = 15
top_recommendations = recommend_movies(user_id=user_id)
print(f"Top 5 recomendações para o usuário {user_id}:")
top_recommendations
```
