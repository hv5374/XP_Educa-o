```python
from tensorflow.keras.layers import Dropout

modelo_com_dropout = Sequential([
    Dense(256, activation='relu', input_shape=(10,)),
    Dropout(0.3), # Remove 30% dos neurônios durante o treinamento
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])

modelo_com_dropout.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
modelo_com_dropout.summary()
```
