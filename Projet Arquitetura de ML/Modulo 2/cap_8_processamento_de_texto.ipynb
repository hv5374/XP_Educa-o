### INDENTIFICAÇÃO DE ENTIDADES

```python
# Carrega o modelo de PLN para português
nlp = spacy.load("pt_core_news_sm")

# Texto com informações sobre um aluno e sua graduação
texto = """
João Pedro concluiu sua graduação na faculdade XPE em São Paulo.
Ele investiu R$ 5 mil no curso e finalizou os estudos em abril de 2024.
"""

# Processar o texto
doc = nlp(texto)

# Exibir entidades reconhecidas
print("Entidades identificadas:\n")
for ent in doc.ents:
 print(f"→ {ent.text:<30} | Tipo: {ent.label_}")
```

```python
for token in doc:
    print(f"{token.text:<15} | {token.pos_:<10} | {token.ent_type_}")
```

```python
for ent in doc.ents:
    print(f"Entidade: {ent.text} | Tipo: {ent.label_}")
```

```python

# Texto com nomes de lugares compostos
texto = """
A sede da empresa está localizada em São Paulo, mas
também temos um escritório no Rio de Janeiro e outro em
Belo Horizonte.
"""

# Processar o texto com spaCy
doc = nlp(texto)

# Lista de entidades do tipo LOC (Localização)
entidades_locais = [ent.text for ent in doc.ents if ent.label_ == "LOC"]

# Criar lista de tokens finais com base nas entidades
tokens_final = []
i = 0

while i < len(doc):
    token_atual = doc[i]
    entidade_encontrada = False

    for ent in doc.ents:
        if ent.label_ == "LOC" and ent.start == i:
            tokens_final.append(ent.text)
            i = ent.end
            entidade_encontrada = True
            break

    if not entidade_encontrada:
        tokens_final.append(token_atual.text)
        i += 1

# Exibir os tokens
print("Tokens finais (com locais agrupados):\n")
print(tokens_final)
```
