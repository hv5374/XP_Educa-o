### Desambiguação ###

```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
# Função de similaridade do cosseno
def similaridade_cosseno(v1, v2):
 return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

# Modelo de embeddings
modelo = SentenceTransformer("all-MiniLM-L6-v2")

# Frases ambíguas com a palavra “banco”
frase_1 = "O cliente foi ao banco para abrir uma conta."
frase_2 = "Ela sentou no banco do parque para descansar."

# Sentidos possíveis
sentido_financeiro = "instituição financeira"
sentido_assento = "assento para sentar"

# Embeddings
vec_frase_1 = modelo.encode(frase_1)
vec_frase_2 = modelo.encode(frase_2)
vec_financeiro = modelo.encode(sentido_financeiro)
vec_assento = modelo.encode(sentido_assento)

# Cálculo das similaridades
print("Frase 1:", frase_1)
print("→ Similaridade com 'instituição financeira':", round(similaridade_cosseno(vec_frase_1, vec_financeiro), 3))
print("→ Similaridade com 'assento para sentar':", round(similaridade_cosseno(vec_frase_1, vec_assento), 3))
print()
print("Frase 2:", frase_2)
print("→ Similaridade com 'instituição financeira':", round(similaridade_cosseno(vec_frase_2, vec_financeiro), 3))
print("→ Similaridade com 'assento para sentar':", round(similaridade_cosseno(vec_frase_2, vec_assento), 3))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Modelo pré-treinado para embeddings semânticos
modelo = SentenceTransformer("all-MiniLM-L6-v2")

# Categorias de interesse (temas comuns)
categorias = [
 "elogio ao atendimento",
 "problema com entrega",
 "qualidade do produto",
 "dúvida sobre pagamento",
 "reclamação sobre preço"
]

# Comentários reais de clientes
comentarios = [
"Fui muito bem atendido pelo suporte, estão de parabéns!",
"O pedido chegou com 5 dias de atraso.",
"O pneu que comprei é de excelente qualidade, gostei bastante.",
"Quero saber se posso pagar com dois cartões.",
"Achei o preço do produto muito alto comparado a outras lojas."
]

# Gerar embeddings dos comentários e categorias
embeddings_comentarios = modelo.encode(comentarios)
embeddings_categorias = modelo.encode(categorias)
# Medir similaridade entre cada comentário e cada categoria
resultados = []
for i, emb_comentario in enumerate(embeddings_comentarios):
 similaridades = cosine_similarity([emb_comentario], embeddings_categorias)[0]
 categoria_mais_proxima = categorias[similaridades.
argmax()]
 resultados.append({
 "Comentário": comentarios[i],
 "Tema Identificado": categoria_mais_proxima,
 "Similaridade": round(similaridades.max(), 2)
 })
# Exibir resultados
df_resultado = pd.DataFrame(resultados)
print(df_resultado)
```
