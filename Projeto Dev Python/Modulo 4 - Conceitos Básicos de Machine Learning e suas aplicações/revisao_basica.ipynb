```python
pessoa = {
    "nome": "Henrique",
    "idade": 40,
    "cidade": "Campina Grande"
}

for i in pessoa:
    print(i, "é", pessoa[i])
```

```python
for chave, valor in pessoa.items():
    print(chave, ":", valor)
```

```python
print(pessoa.keys())
print(pessoa.values())
```

```python
#DICIONÁRIOS DENTRO DE LISTAS#

pessoas = [
    {"nome": "João", "idade": 25},
    {"nome": "Maria", "idade": 30},
    {"nome": "Pedro", "idade": 20}
]

for pessoa in pessoas:
    print("Nome:", pessoa["nome"])
    print("Idade:", pessoa["idade"])
    print()

print(pessoas[0]["nome"])
print(pessoas[0]["idade"])
```
