```python
'''SEM FUNCÃO'''

numero = 10
print(numero * 2)

numero = 20
print(numero * 2)

numero = 50
print(numero * 2)
```

```python
'''COM FUNÇÃO'''
def dobro(numero):
    return numero * 2

print(dobro(10))
print(dobro(20))
print(dobro(50))

```

```python
'''simplesmente definir a função não a executa'''
def saudacao():
    print("Olá!")

'''Precisamos chamá-la:'''
saudacao()
```

```python
#Podemos fazer a função receber uma informação.
def saudacao(nome):         # nome é Parametro
    print("Olá,", nome)

saudacao("Alice")           # Alice é o Argumento
saudacao("Bob")             # Bob é o Argumento
saudacao("Charlie")         # Charlie é o Argumento
```

```python
#Vários parâmetros

#Uma função pode receber vários valores:
def apresentar(nome, idade):
    print("Nome:", nome)
    print("Idade:", idade)


apresentar("Henrique", 40)
apresentar("Alice", 30)
apresentar("Bob", 25)
```

```python
def somar(a, b, c):    # Definindo a função com três parâmetros
    print(a + b + c)

#Podemos fazer:

somar(10, 20, 30)      # Passando os argumentos diretamente
```

```python
#E se quisermos guardar o resultado?

#Usamos return:
def somar(a, b, c):               # Definindo a função com três parâmetros
    return a + b + c              # Retorna a soma dos três valores

# Agora podemos guardar o resultado em uma variável:
resultado = somar(10, 20, 30)     # Guardando o resultado da função na variável 'resultado'
print(resultado)                  # Saída: 60
```

```python
#Por que return é tão importante?

#Porque podemos usar o resultado em outro lugar.
def somar(a, b):
    return a + b

resultado = somar(10, 20)

if resultado > 25:                          # Verificando se o resultado é maior que 25
    print("Resultado maior que 25.")
```
