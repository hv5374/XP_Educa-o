Correção de Dados Ausentes - Variáveis Categóricas

```python
df.isna().sum()                                                          #Mostra a quantidade de valores nulos em cada coluna
```

```python
filtro = df["classe_trabalho"].isna() | df["qtde_filhos"].isna()    #Mostra as linhas que possuem valores nulos em qualquer uma das colunas
index_ausentes = df[filtro].index                                   #Mostra os índices das linhas que possuem valores nulos em qualquer uma das colunas
df.loc[index_ausentes]                                              #Mostra as linhas que possuem valores nulos em qualquer uma das colunas
```

```python
moda = df["classe_trabalho"].mode()[0]  #Calcula a moda da coluna "classe_trabalho"
df["classe_trabalho"] = df["classe_trabalho"].fillna(value=moda)  #Substitui os valores nulos na coluna "classe_trabalho" pela moda
df.loc[index_ausentes]
```

```python
df["qtde_filhos"] = df["qtde_filhos"].fillna(value=0)  #Substitui os valores nulos na coluna "qtde_filhos" por 0
df.loc[index_ausentes]
```
