Avaliar a correlação entre as variáveis

Faça a integração dos dados dos conjuntos de dados

```python
df0 = pd.merge(df0, df1, how='inner', on='id_cliente')
df0 = pd.merge(df0, df2, how='inner', on='id_estado')

df0.head()
```

```python
#Qual é a média de salários das pessoas do gênero masculino e que moram em Minas Gerais?

filtro = df0[
    (df0["genero"] == "Masculino") &
    (df0["estado"] == "Minas Gerais")
]

print(filtro["salario"].mean())
```

```python
#Qual estado civil tem a menor quantidade de filhos e qual é esse número? Realize a soma da quantidade de filhos.
df0.groupby("estado_civil")["qtde_filhos"].sum().sort_values()
```

```python
#Qual é a média salarial das pessoas casadas que possuem ensino superior completo e trabalham como Funcionário de Setor Privado?

filtro = df0[
    (df0["estado_civil"] == "Casado") &
    (df0["escolaridade"] == "Superior Completo") &
    (df0["classe_trabalho"] == "Funcionário Setor Privado")
]

print(filtro["salario"].mean())
```

```python
#Qual é a classe de trabalho que apresenta a maior média de salários e qual é o valor associado para as pessoas do estado de Pernambuco?

df_pe = df0[df0["estado"] == "Pernambuco"]

df_pe.groupby("classe_trabalho")["salario"].mean().sort_values(ascending=False)
```

```python
filtro = df0[(df0["peso"] > 120) & (df0["regiao"] == "Sul")]

media = st.mean(filtro["colesterol"])
desvio_padrao = st.stdev(filtro["colesterol"])

print("Média:", media)
print("Desvio padrão:", desvio_padrao)
```
