Exemplo prático de Normalização e Padronização de dados

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
df = pd.DataFrame(
    {
        'idade': [18, 25, 40, 60, 80], 
        'salario': [1200, 2500, 5000, 8000, 25000]
    }
)

df.head()
```

Normalização

```python
scaler = MinMaxScaler() # Normalização
colunas = ['idade', 'salario']

df[["idade_normalizada", "salario_normalizado"]] = scaler.fit_transform(df[colunas]) 
df.head()
```

Padronização

```python
scaler = StandardScaler() # Padronização
colunas = ['idade', 'salario']

df[["idade_padronizada", "salario_padronizado"]] = scaler.fit_transform(df[colunas]) 
df.head().round(2)
```
